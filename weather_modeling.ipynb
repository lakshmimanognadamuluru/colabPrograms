```python
def Temperature_modeling(a, b, c, time):
    temperature = a * time**2 + b * time + c
    return temperature
```

```python
print("Hard-coded variables for weather modeling")
print("Temperature for hardcoded coefficients at time", time, "hours:", Temperature_modeling(a, b, c, time))
```

```python
def read_from_file(filename):
    with open(filename, 'r') as file:
        # Read each line, split into values, and convert to floats
        return [list(map(float, line.split())) for line in file.readlines()]
```

```python
# Example: Solve for all sets in the file
coefficients = read_from_file('program2.txt')

total_temperature = 0

# Iterate through each set of coefficients
for idx, (a, b, c, time) in enumerate(coefficients):
    temperature = Temperature_modeling(a, b, c, time)  # Assuming Temperature_modeling exists
    total_temperature += temperature
    print(f"Set {idx + 1}: Temperature from file coefficients at time {time} hours: {temperature}")

print(f"Total temperature across all sets: {total_temperature}")
```
